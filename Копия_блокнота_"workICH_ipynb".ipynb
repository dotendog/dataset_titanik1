```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
import pandas as pd
df = pd.read_csv("/content/Titanic-Dataset.csv")
df.head()
```

```python
median_age=df['Age'].median()
df['Age']=df['Age'].fillna(median_age)
df
```

```python
df.drop(
    ['Cabin'],
    axis = 1, #удаление столбца
    inplace = True,
    errors = 'ignore'#игнорировать ошибку, если нет столбца с указанным именем
)
df
```

1) Построить гистограммы (распределение числового признака) для двух
числовых столбцов.

```python
vozrat = np.random.choice(df['Age'], size=890)
plt.figure(figsize=(7, 4))
plt.hist(vozrat, bins=17, color='orchid', edgecolor='black', alpha=0.7)
plt.title('Распределение возраста пассажиров ')
plt.xlabel('Возраст (лет)')
plt.ylabel('Частота')
plt.grid(axis='y', linestyle='--', alpha=0.9)
plt.show()
```

2. Построить столбчатую диаграмму для категориального признака.

```python
fig, axs = plt.subplots(1, 2, figsize=(15, 4))

# столбчатая диаграмма

axs[0].bar(df['Pclass'], df['Age'], color='mediumseagreen')
axs[0].set_title('Распределение пассажиров по классу (столбчатая)')
axs[0].set_xlabel('Класс')
axs[0].set_ylabel('Возраст пассажиров в классе')
axs[0].grid(axis='y', linestyle='--', alpha=0.7)

# линейчатая диаграмма

axs[1].barh(df['Pclass'], df['Age'], color='cornflowerblue')
axs[1].set_title('Распределение пассажиров по классу (линейчатая)')
axs[1].set_xlabel('Возраст пассажиров в классе')
axs[1].set_ylabel('Класс')
axs[1].grid(axis='x', linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()
```

3. Построить круговую диаграмму для категориального признака.

```python
vzil = df['Survived'].value_counts()
labels= ['выжившие', 'не выжившие']
plt.figure(figsize=(6, 6))
plt.pie(
    vzil,
    labels=labels,
    autopct='%1.1f%%',
    startangle=90,
    colors=['skyblue','gold'],
    explode=[0.1, 0]  # "выдвинуть" категорию «мужчина» для наглядности
)
plt.title('Процентное соотношение (не) выживших')
plt.show()
```

```python
# График 1 (класс каюты по ID)
plt.figure(figsize=(12, 5))
plt.plot(df['PassengerId'], df['Pclass'],c='hotpink', linestyle='-', marker='*', markersize=8, label='Класс каюты (Pclass)')
plt.title('Зависимость класса каюты от ID пассажира')
plt.xlabel('ID пассажира ')
plt.ylabel('Класс каюты (1–3)')
plt.legend()
plt.grid(True)
plt.show()

# График 2:  (возраст по ID)
plt.figure(figsize=(12, 5))
plt.plot(df['PassengerId'], df['Age'],color='skyblue', linestyle=':', marker='o', markersize=8, label='Возраст (Age)')
plt.title('Зависимость возраста от ID пассажира')
plt.xlabel('ID пассажира ')
plt.ylabel('Возраст (лет)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
# Выживаемость по полу
survival_sex = df.groupby('Sex')['Survived'].mean() * 100
ax1.bar(
    survival_sex.index,
    survival_sex.values,
    color=['lightcoral', 'lightblue'],
    )
ax1.set_xlabel('Пол')
ax1.set_ylabel('Доля выживших (%)')
ax1.set_title('Выживаемость по полу')
ax1.grid(axis='y', linestyle='--',alpha=0.3)
# Добавляем значения на столбцы
for i, v in enumerate(survival_sex.values):
    ax1.text(i, v + 1, f'{v:.1f}%', ha='center', va='bottom')
# Выживаемость по классу
survival_class = df.groupby('Pclass')['Survived'].mean() * 100
ax2.bar(
    survival_class.index,
    survival_class.values,
    color=['gold', 'green', 'red'],
    )
ax2.set_xlabel('Класс')
ax2.set_ylabel('Доля выживших (%)')
ax2.set_title('Выживаемость по классу')
ax2.grid(axis='y', alpha=0.3)
# Добавляем значения на столбцы
for i, v in enumerate(survival_class.values):
    ax2.text(i + 1, v + 1, f'{v:.1f}%', ha='center', va='bottom')
plt.tight_layout()
plt.show()
```
